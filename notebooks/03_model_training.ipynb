# 03: Model Training & Chronological Validation

Trains Seasonal Naive baseline, SARIMA, and XGBoost regression models on chronologically split train (80%), validation (10%), and test (10%) sets.

```python
import os
import sys
import pandas as pd
sys.path.append('..')
from models.model_utils import chronological_split, calculate_metrics
from models.xgboost_model import RetailXGBoostModel

feat_df = pd.read_parquet('../data/processed/features.parquet')
train_df, val_df, test_df = chronological_split(feat_df, 0.8, 0.1)
print(f"Train: {len(train_df):,} | Val: {len(val_df):,} | Test: {len(test_df):,}")

# Train XGBoost regressor
model = RetailXGBoostModel(n_estimators=200, learning_rate=0.05)
model.fit(train_df, val_df)
metrics = model.evaluate(val_df)
print("XGBoost Validation Results:", metrics)
```
