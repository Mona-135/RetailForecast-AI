# 01: Retail Sales Data Exploration

This notebook loads the Walmart Recruiting Store Sales dataset, analyzes store-department distributions, inspects holiday spikes, and performs baseline data profiling.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load raw Walmart train dataset
train_df = pd.read_csv('../data/raw/train.csv')
stores_df = pd.read_csv('../data/raw/stores.csv')
features_df = pd.read_csv('../data/raw/features.csv')

print(f"Train records: {len(train_df):,}")
train_df.head()
```

```python
# Summary statistics
print("Stores count:", train_df['Store'].nunique())
print("Departments count:", train_df['Dept'].nunique())
print("Date range:", train_df['Date'].min(), "to", train_df['Date'].max())
train_df.describe()
```

```python
# Plot weekly sales over time
train_df['Date'] = pd.to_datetime(train_df['Date'])
weekly_total = train_df.groupby('Date')['Weekly_Sales'].sum()

plt.figure(figsize=(12, 5))
plt.plot(weekly_total.index, weekly_total.values, color='#3b82f6', lw=2)
plt.title('Total Weekly Sales Volume Over Time (Walmart Fleet)')
plt.ylabel('Total Sales ($)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
