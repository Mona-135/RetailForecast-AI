# 02: Chronological Feature Engineering & Lag Generation

Demonstrates how to construct weekly lags (`lag_1`, `lag_2`, `lag_4`, `lag_7`), rolling window statistics, and calendar attributes without future look-ahead data leakage.

```python
import os
import sys
import pandas as pd
sys.path.append('..')
from spark.feature_engineering import generate_features_pandas

# Load cleaned sales data
sales_df = pd.read_parquet('../data/processed/sales_clean.parquet')
feat_df = generate_features_pandas(sales_df)
print(f"Generated {len(feat_df.columns)} features across {len(feat_df):,} records.")
feat_df[['store', 'department', 'date', 'sales', 'lag_1', 'rolling_mean_4', 'rolling_std_4']].head(10)
```

```python
# Correlation matrix of features
numeric_cols = ['sales', 'lag_1', 'lag_2', 'lag_4', 'rolling_mean_4', 'rolling_mean_8', 'rolling_std_4']
corr = feat_df[numeric_cols].corr()
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap='Blues', fmt='.2f')
plt.title('Feature Autocorrelation Matrix')
plt.show()
```
